```python
# Calculate discount for a VIP customer and normal customer
def calculate_discount(bill_amount, vip_member=False):
    if vip_member == True:
        discounted_total = bill_amount * 0.85
    else:
        discounted_total = bill_amount * 0.95
    return discounted_total
vip_bill = calculate_discount(1000, True)
normal_bill = calculate_discount(1000)
print("VIP customer bill:", vip_bill)
print("Normal customer bill:", normal_bill)
```

```python
# Reward Points Calculator
points_earned = lambda amount: amount // 10
# Calculate reward points
print("Reward points:", points_earned(250))
```

```python
# Recursive Tier Bonus Calculator and Test the function
def tier_bonus(level):
    if level == 1:
        return 50
    else:
        return 50 + tier_bonus(level - 1)
result = tier_bonus(3)
print("Total bonus points:", result)
```

```python
#OOP
class StoreBranch:
    # Class variable (shared by all branches)
    company_name = "Urban Retail"
    def __init__(self, branch_id, daily_revenue):
        self.branch_id = branch_id
        self.__daily_revenue = daily_revenue
        # private variable (double underscore)
    def get_revenue(self):
        return self.__daily_revenue
class FlagshipStore(StoreBranch):
    def __init__(self, branch_id, daily_revenue, vip_lounge_access):
        super().__init__(branch_id, daily_revenue)
        # sets up the parent's variables
        self.vip_lounge_access = vip_lounge_access
    # Method overriding: same name as the parent's method, new behaviour
    def get_revenue(self):
        return super().get_revenue() + 500.0
# Create the object and print its final revenue
store = FlagshipStore(101, 2000.0, True)
print("Final revenue:", store.get_revenue())
```
